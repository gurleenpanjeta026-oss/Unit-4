Conditions

```python
a = 13
b = 24
if b > a:
    print(f'(b) is greater than (a)')
```

```python
a = 13
b = 24
if b > a:
    print(f'(b) is greater than (a)')
    print('Hello')
```

```python
a = 13
b = 24
if b > a:
    print(f'(b) is greater than (a)')
else:
    print(f'(b) is less than or equal to (a)')
```

```python
a = 41
b = 85
if b > a:
    print(f'(b) is greater than (a)')
elif b == a:
    print(f'(b) is equal to (a)')
else:
    print(f'(b) is less than (a)')
```

Loops

While

```python
i = 0
while i < 4:
    print(i)
    i+=1
    
```

```python
i = 0
while i <10:
    print(2*i+3)
    i+=3
```

For

```python
cars = ['Tesla', 'Mazda', 'Ford', 'Toyota', 'Benz']
```

```python
for car in cars:
    print(car)
```

```python
for i in cars:
    print(i)
```

```python
for i in range(6):
    print(i)
```

```python
course = "Business Context in Data Analytics"
```

```python
all_letters = []
for letter in course:
    all_letters.append(letter)
```

```python
print(all_letters)
```

```python
print(course.lower())
```

```python
print(course.upper())
```

```python
all_letters_2 = [letter for letter in course]
print(all_letters_2)
```

```python
vowel_letters = 'aeiouAEIOU'
```

```python
course_vowel_letters = [letter for letter in course if letter in vowel_letters]
print(course_vowel_letters)
```

```python
course_non_vowel_letters = [letter for letter in course if letter not in vowel_letters]
print(course_non_vowel_letters)
```

```python
vowel_letters_low = 'aeiou'
```

```python
course_non_vowel_letters_2 = [letter for letter in course.lower() if letter not in vowel_letters_low]
print(course_non_vowel_letters_2)
```
